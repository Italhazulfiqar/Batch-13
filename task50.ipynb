```python
# Task 50: Standard deviation and variance
# Tutorial: https://www.w3schools.com/python/python_ml_standard_deviation.asp

import numpy as np

# The two datasets come from the linked standard-deviation exercise.
close_speeds = np.array([86, 87, 88, 86, 87, 85, 86])
spread_speeds = np.array([32, 111, 138, 28, 59, 77, 97])
for label, speeds in [('Closely grouped speeds', close_speeds),
                      ('Widely spread speeds', spread_speeds)]:
    print(label + ':', speeds)
    print('Mean:', round(float(np.mean(speeds)), 6))
    print('Population standard deviation:', round(float(np.std(speeds, ddof=0)), 6))
    print('Population variance:', round(float(np.var(speeds, ddof=0)), 6))
    print('Square root of variance:', round(float(np.sqrt(np.var(speeds))), 6))

# Verify the population formula manually for the second dataset.
mean = float(np.mean(spread_speeds))
squared_differences = (spread_speeds - mean) ** 2
manual_variance = float(np.sum(squared_differences) / len(spread_speeds))
print('Manual population variance:', round(manual_variance, 6))
print('Sample standard deviation (ddof=1):', round(float(np.std(spread_speeds, ddof=1)), 6))
```
