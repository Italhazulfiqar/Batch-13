```python
# Task 61: Text processing with NLTK
# Reference: https://pythonspot.com/category/nltk/

from pathlib import Path
import nltk

# The download includes these resources, so this setup works offline.
# Running an individual file without the data folder requires internet once.
base = Path(__file__).resolve().parent if '__file__' in globals() else Path.cwd()
resource_folder = base / 'data' / 'nltk_data'
nltk.data.path.insert(0, str(resource_folder))
resources = {
    'corpora/stopwords': 'stopwords',
    'tokenizers/punkt_tab/english/': 'punkt_tab',
    'taggers/averaged_perceptron_tagger_eng/': 'averaged_perceptron_tagger_eng',
    'corpora/wordnet.zip': 'wordnet',
}
for resource_path, package in resources.items():
    try:
        nltk.data.find(resource_path)
    except LookupError:
        if not nltk.download(package, download_dir=str(resource_folder), quiet=True):
            raise RuntimeError('Could not download NLTK resource: ' + package)

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import sent_tokenize, word_tokenize

text = 'Python makes text processing useful. Students are learning and writing programs.'
sentences = sent_tokenize(text)
words = word_tokenize(text)
print('Sentences:', sentences)
print('Word tokens:', words)
stop_words = set(stopwords.words('english'))
cleaned = [word.lower() for word in words if word.isalpha()
           and word.lower() not in stop_words]
print('Stop-word filtering:', cleaned)

stemmer = PorterStemmer()
print('Stemming:')
for word in ['learning', 'programs', 'studies', 'running']:
    print(word, '->', stemmer.stem(word))

lemmatizer = WordNetLemmatizer()
print('Noun lemma for programs:', lemmatizer.lemmatize('programs'))
print('Verb lemma for running:', lemmatizer.lemmatize('running', pos='v'))
print('Part-of-speech tags:')
print(nltk.pos_tag(words))
print('Token frequencies:')
print(nltk.FreqDist(cleaned).most_common())
```
