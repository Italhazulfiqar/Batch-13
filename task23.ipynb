## Example 1: Multiline text


```python
# Multiline string
message = """To avoid pain, they avoid pleasure.
To avoid death, they avoid life."""

print(message)
```

## Example 2: Positive indexing


```python
model = 'ChatGPT'

# Access the first character
print(model[0])    # Output: C

# Access the fifth character
print(model[4])    # Output: G
```

## Example 3: Negative indexing


```python
model = 'ChatGPT'

# Access the last character
print(model[-1])   # Output: T

# Access the fourth last character
print(model[-4])   # Output: t
```

## Example 4: Index outside the string


```python
model = 'ChatGPT'
print(model[8])  
```

## Example 5: Slicing


```python
model = 'ChatGPT'

# Access characters from index 0 up to (but not including) 4
print(model[0:4])   # Output: Chat
```

## Example 6: Strings cannot be modified in place


```python
model = 'ChatGPT'

model[0] = 'W'
print(model)
```

## Example 7: Concatenation and reassignment


```python
model = 'Opus'
version = '5'

model = model + " " + version
print(model)    # Opus 5
```

## Example 8: Replacing text


```python
text = "ChatGPT is great."

# Replace "ChatGPT" with "Claude"
new_text = text.replace("ChatGPT", "Claude")

print(new_text)  # Output: Claude is great.
```

## Example 9: Membership


```python
print('Chat' in 'ChatGPT')        # True
print('Claude' not in 'ChatGPT')  # True
```

## Example 10: Character iteration


```python
model = 'Opus'

for c in model:
    print(c)
```

## Example 11: Length


```python
model = 'Opus'

# Count the number of characters
print(len(model))   # Output: 4
```

## Example 12: Invalid unescaped quotation marks


```python
example = "He said, "What's there?""

print(example) # Error
```

## Example 13: Escaping quotes


```python
# escape double quotes
example = "He said, \"What's there?\""

# escape single quotes
example = 'He said, "What\'s there?"'

print(example)

# Output: He said, "What's there?"
```

## Example 14: Formatting with an f-string


```python
company = 'Google'
field = 'AI'

message = f'{company} is an {field} company.'
print(message)
```
