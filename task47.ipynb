```python
# Task 47: Cleaning pandas data
# Tutorial: https://www.w3schools.com/python/pandas/pandas_cleaning.asp

import pandas as pd

# Deliberately include a missing cell, mixed date formats, a wrong value,
# a duplicate row and an invalid date to demonstrate all four cleaning topics.
df = pd.DataFrame({
    'Duration': [60, 45, 450, 30, 45, 30],
    'Date': ['2020/12/01', '20201202', '2020/12/03', '2020/12/04',
             '20201202', 'invalid'],
    'Pulse': [110, 105, 100, 95, 105, 90],
    'Calories': [409.1, 282.4, None, 195.1, 282.4, 180.0]
})
print('Original data:')
print(df.to_string(index=False))
print('Missing values:')
print(df.isna().sum().to_string())
print('Rows after dropping empty cells:', len(df.dropna()))

cleaned = df.copy()
# Fill empty numeric cells with the mean.
mean_calories = cleaned['Calories'].mean()
cleaned['Calories'] = cleaned['Calories'].fillna(mean_calories)
print('Mean used for empty Calories:', round(float(mean_calories), 2))

# Parse both supported formats; invalid text becomes NaT.
cleaned['Date'] = pd.to_datetime(cleaned['Date'], format='mixed', errors='coerce')
print('Invalid dates:', int(cleaned['Date'].isna().sum()))
cleaned = cleaned.dropna(subset=['Date']).copy()

# Values over 120 minutes are treated as input errors in this exercise.
wrong_duration = cleaned['Duration'] > 120
print('Wrong duration values:', int(wrong_duration.sum()))
cleaned.loc[wrong_duration, 'Duration'] = 60
print('Duplicate rows:', int(cleaned.duplicated().sum()))
cleaned = cleaned.drop_duplicates().reset_index(drop=True)
print('Cleaned data:')
print(cleaned.to_string(index=False))
print('Remaining missing cells:', int(cleaned.isna().sum().sum()))
print('Remaining duplicate rows:', int(cleaned.duplicated().sum()))
```
