## Example 1: Iterating through a list


```python
a = ["Geeks", "for", "Geeks"]
for i in a:
    print(i)
```

## Example 2: Iterating through characters


```python
s = "Geeks"
for i in s:
    print(i)

```

## Example 3: A range with a step


```python
for i in range(0, 10, 2):
    print(i)

```

## Example 4: Skipping letters with continue


```python
for i in 'geeksforgeeks':
    if i == 'e' or i == 's':
        continue
    print(i)
```

## Example 5: Stopping with break


```python
for letter in 'training':
    if letter == 'i':
        break
print(letter)


```

## Example 6: Using pass as a placeholder


```python
for i in 'geeksforgeeks':
    pass
```

## Example 7: Else runs after normal completion


```python
for i in range(1, 4):
    print(i)
else:  
    print("No Break\n")
```

## Example 8: Getting indexes with enumerate


```python
b = ["eat", "sleep", "repeat"]
for i, j in enumerate(b):
    print (i, j)
```

## Example 9: Nested loops


```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```
