```python
# Task 48: Pandas correlations
# Tutorial: https://www.w3schools.com/python/pandas/pandas_correlations.asp

import pandas as pd

# Sample numerical data, plus a text column excluded from correlation.
df = pd.DataFrame({
    'Session': ['A', 'B', 'C', 'D', 'E', 'F'],
    'Duration': [20, 30, 40, 50, 60, 70],
    'Pulse': [100, 108, 102, 115, 110, 112],
    'Calories': [120, 180, 240, 300, 360, 420],
    'RestTime': [70, 60, 50, 40, 30, 20]
})
print('Sample data:')
print(df.to_string(index=False))
correlation = df.corr(numeric_only=True)
print('Pearson correlation matrix:')
print(correlation.round(3).to_string())
print('Duration vs Calories:', round(float(correlation.loc['Duration', 'Calories']), 3))
print('Duration vs RestTime:', round(float(correlation.loc['Duration', 'RestTime']), 3))
# A correlation measures association; it does not establish causation.
```
