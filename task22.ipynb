## Example 1: Empty, integer, mixed, nested, packed and unpacked tuples


```python

# empty tuple
my_tuple = ()

# tuple having integers
my_tuple = (1, 2, 3)

# tuple with mixed datatypes
my_tuple = (1, "Hello", 3.4)

# nested tuple
my_tuple = ("mouse", [8, 4, 6], (1, 2, 3))

# tuple can be created without parentheses
# also called tuple packing
my_tuple = 3, 4.6, "dog"
# tuple unpacking is also possible
a, b, c = my_tuple
```

## Example 2: Single-element tuple needs a comma


```python
my_tuple = ("hello")  # Only parentheses is not enough
print(type(my_tuple))

my_tuple = ("hello",)  # Need a comma at the end
print(type(my_tuple))

my_tuple = "hello",  # Parentheses are optional
print(type(my_tuple))
```

## Example 3: Indexing and index errors and nested indexes


```python
my_tuple = ['p', 'e', 'r', 'm', 'i', 't']

print(my_tuple[0])
print(my_tuple[5])

try:
    print(my_tuple[6])  # Index must be in range
except IndexError as e:
    print("IndexError:", e)

try:
    print(my_tuple[2.0])  # Index must be an integer
except TypeError as e:
    print("TypeError:", e)

n_tuple = ("mouse", [8, 4, 6], (1, 2, 3))

print(n_tuple[0][3])  # Nested index
print(n_tuple[1][1])  # Nested index
print(n_tuple[2][0])  # Nested index
```

## Example 4: Negative indexes


```python
my_tuple = ['p', 'e', 'r', 'm', 'i', 't']

print(my_tuple[-1])
print(my_tuple[-6])
```

## Example 5: Slices


```python
my_tuple = ('p', 'r', 'o', 'g', 'r', 'a', 'm', 'i', 'z')

print(my_tuple[1:4])   # Elements 2nd to 4th
print(my_tuple[:-7])   # Elements beginning to 2nd
print(my_tuple[7:])    # Elements 8th to end
print(my_tuple[:])     # Elements beginning to end
```

## Example 6: Immutable tuple with a mutable nested list


```python
my_tuple = (4, 2, 3, [6, 5])

try:
    my_tuple[1] = 9  # We cannot change an element
except TypeError as e:
    print("TypeError:", e)

try:
    my_tuple[3] = 9  # We cannot change an element
except TypeError as e:
    print("TypeError:", e)

my_tuple[3][0] = 9  # Item of mutable element can be changed
print(my_tuple)

my_tuple = ('p', 'r', 'o', 'g', 'r', 'a', 'm', 'i', 'z')  # Tuples can be reassigned
print(my_tuple)
```

## Example 7: Concatenation and repetition


```python
print((1, 2, 3) + (4, 5, 6))  # Tuple concatenation

print(("Repeat",) * 3)  # Tuple repetition
```

## Example 8: Deletion errors and deleting a whole tuple


```python
my_tuple = ('p', 'r', 'o', 'g', 'r', 'a', 'm', 'i', 'z')

try:
    del my_tuple[3]  # Can't delete individual items
except TypeError as e:
    print("TypeError:", e)

del my_tuple  # Can delete the entire tuple

try:
    print(my_tuple)
except NameError as e:
    print("NameError:", e)
```

## Example 9: Count and index


```python
my_tuple = ('a', 'p', 'p', 'l', 'e')

print(my_tuple.count('p'))  # Count occurrences of 'p'
print(my_tuple.index('l'))  # Find index of 'l'
```

## Example 10: Membership


```python
my_tuple = ('a', 'p', 'p', 'l', 'e')

print('a' in my_tuple)
print('b' in my_tuple)
print('g' not in my_tuple)
```
