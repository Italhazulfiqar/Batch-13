```python
# Task 46: Analyzing pandas data
# Tutorial: https://www.w3schools.com/python/pandas/pandas_analyzing.asp

from io import StringIO
import pandas as pd

# A small example dataset includes one missing calorie value.
csv_data = 'Duration,Pulse,Maxpulse,Calories\n60,110,130,409.1\n45,105,125,282.4\n30,100,120,\n60,115,140,430.2\n45,108,132,320.0\n30,95,115,190.5\n60,112,138,410.0\n'
df = pd.read_csv(StringIO(csv_data))
print('First ten rows (or all rows if fewer):')
print(df.head(10).to_string())
print('Default head (five rows):')
print(df.head().to_string())
print('Default tail (five rows):')
print(df.tail().to_string())
print('DataFrame information:')
df.info()
print('Missing values per column:')
print(df.isna().sum().to_string())
print('Summary statistics:')
print(df.describe().round(2).to_string())
```
