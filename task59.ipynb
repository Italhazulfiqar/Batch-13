```python
# Task 59: Seven neural-network loss functions
# Reference: https://www.analyticsvidhya.com/blog/2019/08/detailed-guide-7-loss-functions-machine-learning-python-code/

import numpy as np

# Regression losses.
actual = np.array([3., -1., 2., 7.])
predicted = np.array([2.5, 0., 2.2, 5.])
errors = actual - predicted
mse = np.mean(errors ** 2)
mae = np.mean(np.abs(errors))
delta = 1.0
absolute = np.abs(errors)
huber = np.mean(np.where(absolute <= delta, 0.5 * errors ** 2,
                         delta * (absolute - 0.5 * delta)))
print('Mean squared error:', round(float(mse), 6))
print('Mean absolute error:', round(float(mae), 6))
print('Huber loss (delta=1):', round(float(huber), 6))

# Binary cross-entropy requires probabilities; clip away from log(0).
binary_truth = np.array([1., 0., 1., 0.])
binary_prob = np.clip(np.array([0.9, 0.2, 0.7, 0.1]), 1e-8, 1 - 1e-8)
bce = -np.mean(binary_truth * np.log(binary_prob) +
               (1 - binary_truth) * np.log(1 - binary_prob))
print('Binary cross-entropy:', round(float(bce), 6))

# Hinge loss uses class labels -1 or +1 and raw decision scores.
hinge_labels = np.array([1., -1., 1., -1.])
scores = np.array([0.8, -0.3, 1.4, 0.2])
hinge = np.mean(np.maximum(0, 1 - hinge_labels * scores))
print('Hinge loss:', round(float(hinge), 6))

# Categorical cross-entropy for one-hot multiclass targets.
one_hot = np.array([[1., 0., 0.], [0., 1., 0.], [0., 0., 1.]])
probabilities = np.array([[0.8, 0.1, 0.1], [0.2, 0.7, 0.1], [0.1, 0.2, 0.7]])
categorical = -np.mean(np.sum(one_hot * np.log(probabilities), axis=1))
print('Categorical cross-entropy:', round(float(categorical), 6))

# KL divergence compares two probability distributions.
p = np.array([0.5, 0.3, 0.2])
q = np.array([0.4, 0.4, 0.2])
kl = np.sum(p * np.log(p / q))
print('KL divergence:', round(float(kl), 6))
```
