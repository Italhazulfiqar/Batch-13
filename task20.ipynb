## Example 3: Arguments and a returned result


```python
def evenOdd(x):
    if (x % 2 == 0):
        return "Even"
    else:
        return "Odd"

print(evenOdd(16))
print(evenOdd(7))
```

## Example 4: Default arguments


```python
def myFun(x, y=50):
    print("x: ", x)
    print("y: ", y)

myFun(10)
```

## Example 5: Keyword arguments


```python
def student(fname, lname):
    print(fname, lname)

student(fname='Geeks', lname='Practice')
student(lname='Practice', fname='Geeks')
```

## Example 6: Positional arguments depend on order


```python
def nameAge(name, age):
    print("Hi, I am", name)
    print("My age is ", age)

print("Case-1:")
nameAge("Olivia", 27)

print("Case-2:")
nameAge(27, "Olivia")
```

## Example 7: Arbitrary positional and keyword arguments


```python
def myFun(*args, **kwargs):
    print("Non-Keyword Arguments (*args):")
    for arg in args:
        print(arg)

    print("Keyword Arguments (**kwargs):")
    for key, value in kwargs.items():
        print(f"{key} == {value}")

myFun('Hey', 'Welcome', first='Geeks', mid='for', last='Geeks')
```

## Example 8: An inner function accesses the outer scope


```python
def f1():
    s = 'I love GeeksforGeeks'
    def f2():
        print(s)
        
    f2()
f1()
```

## Example 9: Return a square


```python
def sq_value(num):
    return num**2

print(sq_value(2))
print(sq_value(-4))
```

## Example 10: Mutating a list versus rebinding an integer


```python
def myFun(x):
    x[0] = 20

b = [10, 11, 12, 13]
myFun(b)
print(b)

def myFun2(x):
    x = 20

a = 10
myFun2(a)
print(a)
```
