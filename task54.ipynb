```python
# Task 54: SVM and kernel SVM
# Reference: https://stackabuse.com/implementing-svm-and-kernel-svm-with-pythons-scikit-learn/

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix

iris = load_iris()
x_train, x_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.30, stratify=iris.target, random_state=13)
print('Training/test samples:', len(x_train), len(x_test))
for kernel in ['linear', 'rbf', 'poly']:
    # Scaling is fitted on training data only through the pipeline.
    model = make_pipeline(StandardScaler(), SVC(kernel=kernel, C=1.0, gamma='scale'))
    model.fit(x_train, y_train)
    predictions = model.predict(x_test)
    print('Kernel:', kernel)
    print('Accuracy:', round(float(accuracy_score(y_test, predictions)), 4))
    print('Support vectors per class:', model.named_steps['svc'].n_support_)
    print('Confusion matrix:')
    print(confusion_matrix(y_test, predictions))
```
