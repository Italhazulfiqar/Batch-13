```python
# Task 55: Time-series analysis with pandas
# Reference: https://www.dataquest.io/blog/tutorial-time-series-analysis-with-pandas

import numpy as np
import pandas as pd

# An illustrative daily series with a trend and weekly seasonality.
dates = pd.date_range('2026-01-01', periods=60, freq='D')
index = np.arange(len(dates))
values = 100 + 0.5 * index + 8 * np.sin(2 * np.pi * index / 7)
df = pd.DataFrame({'Consumption': values}, index=dates)
df.index.name = 'Date'
print('Index type:', type(df.index).__name__)
print('First five days:')
print(df.head().round(2).to_string())
print('Date-based selection:')
print(df.loc['2026-01-10':'2026-01-12'].round(2).to_string())
print('Monthly averages:')
print(df.resample('MS').mean().round(2).to_string())
print('Weekly averages (first three weeks):')
print(df.resample('W').mean().head(3).round(2).to_string())

df['Rolling7'] = df['Consumption'].rolling(7, min_periods=7).mean()
df['Lag1'] = df['Consumption'].shift(1)
df['Difference'] = df['Consumption'].diff()
print('Rolling mean, lag and difference:')
print(df.iloc[6:10].round(2).to_string())

# Forecast the final seven days using the previous observed value.
train = df['Consumption'].iloc[:-7]
test = df['Consumption'].iloc[-7:]
predictions = df['Consumption'].shift(1).loc[test.index]
mae = float(np.mean(np.abs(test - predictions)))
print('Chronological training/test sizes:', len(train), len(test))
print('One-step persistence forecast MAE:', round(mae, 4))
```
