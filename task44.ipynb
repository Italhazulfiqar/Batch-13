```python
# Task 44: Pandas DataFrames
# Tutorial: https://www.w3schools.com/python/pandas/pandas_dataframes.asp

from io import StringIO
import pandas as pd

values = {'Calories': [410, 360, 395], 'Duration': [50, 40, 45]}
df = pd.DataFrame(values, dtype='int64')
print('DataFrame:')
print(df.to_string())
print('One row using loc:')
print(df.loc[0].to_string())
print('Two rows using loc:')
print(df.loc[[0, 1]].to_string())

named = pd.DataFrame(values, index=['day1', 'day2', 'day3'], dtype='int64')
print('Named indexes:')
print(named.to_string())
print('Named row day2:')
print(named.loc['day2'].to_string())

csv_data = 'Calories,Duration\n410,50\n360,40\n395,45\n'
loaded = pd.read_csv(StringIO(csv_data))
print('Loaded from embedded CSV:')
print(loaded.to_string())
```
