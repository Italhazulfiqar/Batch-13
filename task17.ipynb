Original practice examples. Run cells in order. Saved outputs appear below code cells and are also preserved as comments. Deliberate errors are handled.


## Example 1: Search and stop when found


```python
a = [1, 3, 5, 7, 9, 11]
val = 7

for i in a:
    if i == val:
        print(f"Found at {i}!")
        break
else:
    print(f"not found")
```

## Example 2: Break in a for loop


```python
for i in range(10):
    print(i)
    if i == 6:
        break
```

## Example 3: Break in a while loop


```python
c = 5

while True:
    print(c)
    c -= 1
    if c == 0:
        print("Countdown finished!")
        break  # Exit the loop
```

## Example 4: Leaving two nested loops with a flag


```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
val = 5
found = False

for r in matrix:
    for n in r:
        if n == val:
            print(f"{val} found!")
            found = True
            break  # Exit the inner loop
    if found:
        break  # Exit the outer loop
```
