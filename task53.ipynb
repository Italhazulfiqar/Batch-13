```python
# Task 53: Decision-tree classification
# Reference: https://www.datacamp.com/community/tutorials/decision-tree-classification-python

import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.metrics import accuracy_score, confusion_matrix

iris = load_iris()
x_train, x_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.30, stratify=iris.target, random_state=13)
print('Features:', list(iris.feature_names))
print('Training/test samples:', len(x_train), len(x_test))
for criterion in ['gini', 'entropy']:
    tree = DecisionTreeClassifier(criterion=criterion, max_depth=3, random_state=13)
    tree.fit(x_train, y_train)
    predictions = tree.predict(x_test)
    print('Criterion:', criterion)
    print('Accuracy:', round(float(accuracy_score(y_test, predictions)), 4))
    print('Tree depth:', tree.get_depth())
    print('Leaves:', tree.get_n_leaves())
    print('Feature importances:', np.round(tree.feature_importances_, 3))
    print('Confusion matrix:')
    print(confusion_matrix(y_test, predictions))

print('Rules from the entropy tree:')
print(export_text(tree, feature_names=list(iris.feature_names), decimals=2))
stump = DecisionTreeClassifier(max_depth=1, random_state=13).fit(x_train, y_train)
print('Decision-stump accuracy:', round(float(stump.score(x_test, y_test)), 4))
```
