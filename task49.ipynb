```python
# Task 49: Mean, median and mode
# Tutorial: https://www.w3schools.com/python/python_ml_mean_median_mode.asp

import numpy as np
from scipy import stats

# Use the speed data supplied in the linked exercise.
speeds = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]
print('Speeds:', speeds)
print('Sorted speeds:', sorted(speeds))
print('Mean:', round(float(np.mean(speeds)), 6))
print('Median:', float(np.median(speeds)))
mode_result = stats.mode(speeds, keepdims=False)
print('Mode:', int(mode_result.mode))
print('Mode frequency:', int(mode_result.count))

# Removing 111 gives the tutorial example with an even number of values.
even_speeds = [99, 86, 87, 88, 86, 103, 87, 94, 78, 77, 85, 86]
print('Median with an even number of values:', float(np.median(even_speeds)))
```
