```python
# Task 43: Pandas Series
# Tutorial: https://www.w3schools.com/python/pandas/pandas_series.asp

import pandas as pd

series = pd.Series([4, 9, 2], dtype='int64')
print('Series with default labels:')
print(series.to_string())
print('First value:', series.iloc[0])

labeled = pd.Series([4, 9, 2], index=['a', 'b', 'c'], dtype='int64')
print('Series with custom labels:')
print(labeled.to_string())
print('Value at label b:', labeled.loc['b'])

calories = {'day1': 410, 'day2': 360, 'day3': 395}
from_dict = pd.Series(calories, dtype='int64')
print('Series from a dictionary:')
print(from_dict.to_string())
print('Selected dictionary keys:')
print(pd.Series(calories, index=['day1', 'day3'], dtype='int64').to_string())

frame = pd.DataFrame({'Calories': from_dict,
                      'Duration': pd.Series([50, 40, 45], index=from_dict.index, dtype='int64')})
print('DataFrame from Series:')
print(frame.to_string())
```
