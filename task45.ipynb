```python
# Task 45: Reading CSV files with pandas
# Tutorial: https://www.w3schools.com/python/pandas/pandas_csv.asp

from pathlib import Path
from tempfile import TemporaryDirectory
import pandas as pd

# Create and read a real CSV in a temporary directory; no missing data.csv.
csv_data = 'Duration,Pulse,Maxpulse,Calories\n60,110,130,409.1\n45,105,125,282.4\n30,100,120,195.1\n60,115,140,430.2\n'
with TemporaryDirectory() as directory:
    csv_path = Path(directory) / 'data.csv'
    csv_path.write_text(csv_data, encoding='utf-8')
    df = pd.read_csv(csv_path)

print('DataFrame from CSV:')
print(df.to_string())
print('Row count:', len(df))
print('Column names:', list(df.columns))
print('Default maximum displayed rows:', pd.options.display.max_rows)
# option_context restores the previous display setting when it exits.
with pd.option_context('display.max_rows', 2):
    print('Display with a limit of two rows:')
    print(df)
with pd.option_context('display.max_rows', 9999):
    print('Display with an increased limit:')
    print(df)
```
