```python
# Task 51: Linear regression
# Reference: https://stackabuse.com/linear-regression-in-python-with-scikit-learn/

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Small study-hours example; values are illustrative, not a real survey.
hours = np.array([1, 1.5, 2, 2.5, 3, 3.5, 4, 4.5, 5, 5.5,
                  6, 6.5, 7, 7.5, 8, 8.5, 9, 9.5, 10, 10.5])
scores = np.array([15, 18, 23, 29, 34, 35, 42, 45, 53, 54,
                   60, 65, 68, 76, 78, 85, 87, 92, 96, 99])
x = hours.reshape(-1, 1)
x_train, x_test, y_train, y_test = train_test_split(
    x, scores, test_size=0.25, random_state=13)
model = LinearRegression().fit(x_train, y_train)
predictions = model.predict(x_test)
print('Training examples:', len(x_train))
print('Test examples:', len(x_test))
print('Slope:', round(float(model.coef_[0]), 4))
print('Intercept:', round(float(model.intercept_), 4))
print('Predicted score for 6 hours:', round(float(model.predict([[6]])[0]), 2))
print('MAE:', round(float(mean_absolute_error(y_test, predictions)), 4))
print('MSE:', round(float(mean_squared_error(y_test, predictions)), 4))
print('R-squared:', round(float(r2_score(y_test, predictions)), 4))
for hour, actual, predicted in zip(x_test.ravel(), y_test, predictions):
    print(f'Hours={hour:.1f}, actual={actual}, predicted={predicted:.2f}')

# A second example uses two independent features.
practice = np.array([2, 1, 3, 2, 4, 3, 5, 4, 6, 5,
                     7, 6, 8, 7, 9, 8, 10, 9, 11, 10])
x_multi = np.column_stack([hours, practice])
y_multi = 10 + 6 * hours + 2 * practice
x_train, x_test, y_train, y_test = train_test_split(
    x_multi, y_multi, test_size=0.25, random_state=13)
multiple = LinearRegression().fit(x_train, y_train)
print('Multiple-regression coefficients:', np.round(multiple.coef_, 4))
print('Multiple-regression prediction:', round(float(multiple.predict([[6, 7]])[0]), 2))
```
