```python
# Task 41: Pandas basics
# Tutorial: https://www.w3schools.com/python/pandas_tutorial.asp

from io import StringIO
import pandas as pd

# Embedded CSV keeps the example runnable without an external data file.
csv_data = 'Name,Marks\nAli,80\nSara,95\nUsman,88\n'
df = pd.read_csv(StringIO(csv_data))
print('DataFrame loaded from CSV:')
print(df.to_string(index=False))
print('Marks column:')
print(df['Marks'].to_string(index=False))
print('Students with marks above 85:')
print(df.loc[df['Marks'] > 85].to_string(index=False))
print('Average marks:', round(float(df['Marks'].mean()), 2))
```
