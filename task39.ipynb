## Example 1: Sort numbers


```python
import numpy as np

arr = np.array([3, 2, 0, 1])

print(np.sort(arr))
```

## Example 2: Sort strings


```python
import numpy as np

arr = np.array(['banana', 'cherry', 'apple'])

print(np.sort(arr))
```

## Example 3: Sort booleans


```python
import numpy as np

arr = np.array([True, False, True])

print(np.sort(arr))
```

## Example 4: Sort each matrix row


```python
import numpy as np

arr = np.array([[3, 2, 4], [5, 0, 1]])

print(np.sort(arr))
```
