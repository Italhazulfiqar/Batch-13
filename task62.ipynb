```python
# Task 62: Text analysis
# Reference: https://medium.com/towards-artificial-intelligence/natural-language-processing-nlp-with-python-tutorial-for-beginners-1f54e610a1a0

from pathlib import Path
import nltk

# The download includes these resources, so this setup works offline.
# Running an individual file without the data folder requires internet once.
base = Path(__file__).resolve().parent if '__file__' in globals() else Path.cwd()
resource_folder = base / 'data' / 'nltk_data'
nltk.data.path.insert(0, str(resource_folder))
resources = {
    'corpora/stopwords': 'stopwords',
    'tokenizers/punkt_tab/english/': 'punkt_tab',
    'taggers/averaged_perceptron_tagger_eng/': 'averaged_perceptron_tagger_eng',
    'corpora/wordnet.zip': 'wordnet',
}
for resource_path, package in resources.items():
    try:
        nltk.data.find(resource_path)
    except LookupError:
        if not nltk.download(package, download_dir=str(resource_folder), quiet=True):
            raise RuntimeError('Could not download NLTK resource: ' + package)

import numpy as np
from nltk.corpus import stopwords
from nltk.tokenize import sent_tokenize, word_tokenize
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text = 'Python supports data analysis. Python supports machine learning. Data helps learning.'
words = [word.lower() for word in word_tokenize(text) if word.isalpha()]
stop_words = set(stopwords.words('english'))
content_words = [word for word in words if word not in stop_words]
print('Sentence count:', len(sent_tokenize(text)))
print('Word count:', len(words))
print('Content-word frequencies:', nltk.FreqDist(content_words).most_common())
print('Content bigrams:', list(nltk.bigrams(content_words)))

documents = ['python data analysis', 'python machine learning', 'data analysis tools']
counts = CountVectorizer().fit(documents)
print('Bag-of-words vocabulary:', list(counts.get_feature_names_out()))
print('Bag-of-words matrix:')
print(counts.transform(documents).toarray())
tfidf = TfidfVectorizer().fit(documents)
vectors = tfidf.transform(documents)
print('TF-IDF matrix:')
print(np.round(vectors.toarray(), 3))
print('Document cosine similarities:')
print(np.round(cosine_similarity(vectors), 3))
```
