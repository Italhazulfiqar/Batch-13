## Example 1: Random integer


```python
from numpy import random

x = random.randint(100)

print(x)
```

## Example 2: Random float


```python
from numpy import random

x = random.rand()

print(x)
```

## Example 3: Random integer vector


```python
from numpy import random

x=random.randint(100, size=(5))

print(x)
```

## Example 4: Random integer matrix


```python
from numpy import random

x = random.randint(100, size=(3, 5))

print(x)
```

## Example 5: Random float vector


```python
from numpy import random

x = random.rand(5)

print(x)
```

## Example 6: Random float matrix


```python
from numpy import random

x = random.rand(3, 5)

print(x)
```

## Example 7: Choose one value


```python
from numpy import random

x = random.choice([3, 5, 7, 9])

print(x)
```

## Example 8: Choose a matrix of values


```python
from numpy import random

x = random.choice([3, 5, 7, 9], size=(3, 5))

print(x)
```
